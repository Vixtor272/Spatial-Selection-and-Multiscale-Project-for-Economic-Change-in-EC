## STEP J — are selection and transmission additive over time?

**Questions to test** (everything in the engine's convention — initial-firm-share weights, relative fitness $w/\bar W$, raw i.e. unannualised):

$$\text{(Q1)}\quad S(19{\to}20) + S(20{\to}21) \overset{?}{=} S(19{\to}21) \qquad\qquad \text{(Q2)}\quad T(19{\to}20) + T(20{\to}21) \overset{?}{=} T(19{\to}21)$$

**Answers: neither holds in general — they fail by the *same* cross term with opposite signs, so their *sum* is exactly additive.** The full proof follows; the code cell below then verifies every statement symbolically (sympy, fully general firm counts and traits) and numerically on the data.

---

### Setup and notation

Fix a common set of parishes $i=1,\dots,n$ present in all three years (firm counts $n_i(t)>0$). For year $t \in \{19, 20, 21\}$ write

$$N_t=\sum_i n_i(t),\qquad p_i(t)=\frac{n_i(t)}{N_t},\qquad z_i(t)=\ln y_i(t),\qquad \langle x\rangle_t := \sum_i p_i(t)\,x_i \;\;\text{(mean of any vector $x$ under year-$t$ firm shares).}$$

For a window $s\to t$ define fitness, mean fitness, selection and transmission:

$$w_i=\frac{n_i(t)}{n_i(s)},\qquad \bar W=\sum_i p_i(s)\,w_i,\qquad S(s{\to}t)=\sum_i p_i(s)\Big(\frac{w_i}{\bar W}-1\Big)\big(z_i(s)-\langle z_s\rangle_s\big),\qquad T(s{\to}t)=\sum_i p_i(t)\,\big(z_i(t)-z_i(s)\big).$$

Abbreviate $S_1=S(19{\to}20)$, $S_2=S(20{\to}21)$, $S_{12}=S(19{\to}21)$, likewise $T_1,T_2,T_{12}$; write $\Delta z_1 := z_{20}-z_{19}$ (the vector of first-period trait changes) and $w_2,\bar W_2$ for the second window's fitness.

---

### Lemma (reweighting). For any window $s\to t$:

**(a)** $\;p_i(t)=p_i(s)\,\dfrac{w_i}{\bar W}\;$ — final shares are initial shares tilted by relative fitness.

*Proof.* First, $\bar W=\sum_i \frac{n_i(s)}{N_s}\cdot\frac{n_i(t)}{n_i(s)}=\frac{N_t}{N_s}$. Then $p_i(s)\,\frac{w_i}{\bar W}=\frac{n_i(s)}{N_s}\cdot\frac{n_i(t)}{n_i(s)}\cdot\frac{N_s}{N_t}=\frac{n_i(t)}{N_t}=p_i(t)$. $\square$

**(b)** For **any fixed vector** $x$ (not just the trait): $\;\operatorname{cov}_{p(s)}\!\Big(\dfrac{w}{\bar W},\,x\Big) \;=\; \langle x\rangle_t-\langle x\rangle_s.$

*Proof.* The weights $p_i(s)\frac{w_i}{\bar W}$ sum to 1 by (a), so
$\operatorname{cov}_{p(s)}\big(\frac{w}{\bar W},x\big)=\sum_i p_i(s)\frac{w_i}{\bar W}x_i-\Big(\underbrace{\textstyle\sum_i p_i(s)\frac{w_i}{\bar W}}_{=1}\Big)\Big(\sum_i p_i(s)x_i\Big)=\sum_i p_i(t)x_i-\langle x\rangle_s=\langle x\rangle_t-\langle x\rangle_s.$ $\square$

**(c)** Applying (b) with $x=z_s$, and rewriting $T$ directly from its definition:

$$S(s{\to}t)=\langle z_s\rangle_t-\langle z_s\rangle_s \qquad\qquad T(s{\to}t)=\langle z_t\rangle_t-\langle z_s\rangle_t.$$

In words: *selection re-weights the frozen initial trait from initial to final shares; transmission updates the trait under frozen final shares.* (Adding them gives the Price identity $S+T=\langle z_t\rangle_t-\langle z_s\rangle_s$ — the two ends of a telescope.)

---

### Step 1 — the long window composes exactly

$w_i^{(19\to21)}=\frac{n_i(21)}{n_i(19)}=w_{1,i}\,w_{2,i}$ and $\bar W_{12}=\frac{N_{21}}{N_{19}}=\frac{N_{20}}{N_{19}}\cdot\frac{N_{21}}{N_{20}}=\bar W_1\bar W_2$. So by Lemma (a) applied twice, $p(21)=p(20)\frac{w_2}{\bar W_2}=p(19)\frac{w_1w_2}{\bar W_1\bar W_2}$ — the endpoint weights are the same whether you take one long step or two short ones. Everything below is bookkeeping with Lemma (c).

### Step 2 — write all six terms as mean differences (Lemma (c))

$$S_1=\langle z_{19}\rangle_{20}-\langle z_{19}\rangle_{19}\qquad S_2=\langle z_{20}\rangle_{21}-\langle z_{20}\rangle_{20}\qquad S_{12}=\langle z_{19}\rangle_{21}-\langle z_{19}\rangle_{19}$$

$$T_1=\langle z_{20}\rangle_{20}-\langle z_{19}\rangle_{20}\qquad T_2=\langle z_{21}\rangle_{21}-\langle z_{20}\rangle_{21}\qquad T_{12}=\langle z_{21}\rangle_{21}-\langle z_{19}\rangle_{21}$$

### Step 3 — Theorem 1 (selection): $\;S_{12}=S_1+S_2-C$, with $\;C:=\operatorname{cov}_{p(20)}\!\big(\tfrac{w_2}{\bar W_2},\,\Delta z_1\big)$

*Proof.* Subtract, cancel the common $\langle z_{19}\rangle_{19}$, and group by trait vector:

$$S_{12}-S_1-S_2=\underbrace{\big(\langle z_{19}\rangle_{21}-\langle z_{19}\rangle_{20}\big)}_{=\,\operatorname{cov}_{p(20)}(w_2/\bar W_2,\;z_{19})\ \text{by (b)}}\;-\;\underbrace{\big(\langle z_{20}\rangle_{21}-\langle z_{20}\rangle_{20}\big)}_{=\,\operatorname{cov}_{p(20)}(w_2/\bar W_2,\;z_{20})\ \text{by (b)}}\;=\;-\operatorname{cov}_{p(20)}\!\Big(\frac{w_2}{\bar W_2},\,z_{20}-z_{19}\Big)=-C,$$

using bilinearity of the covariance in its second argument. $\blacksquare$

### Step 4 — Theorem 2 (transmission): $\;T_{12}=T_1+T_2+C$ — the **same** $C$, opposite sign

*Proof.* In $T_{12}-T_1-T_2$ the term $\langle z_{21}\rangle_{21}$ cancels, leaving

$$T_{12}-T_1-T_2=\big(\langle z_{20}\rangle_{21}-\langle z_{20}\rangle_{20}\big)-\big(\langle z_{19}\rangle_{21}-\langle z_{19}\rangle_{20}\big)=\operatorname{cov}_{p(20)}\!\Big(\frac{w_2}{\bar W_2},\,z_{20}\Big)-\operatorname{cov}_{p(20)}\!\Big(\frac{w_2}{\bar W_2},\,z_{19}\Big)=+C. \;\blacksquare$$

### Step 5 — Corollary: the total is exactly additive

$$\big(S_{12}+T_{12}\big)=\big(S_1+T_1\big)+\big(S_2+T_2\big)=\langle z_{21}\rangle_{21}-\langle z_{19}\rangle_{19}.$$

The cross term $C$ never leaks out of the decomposition — it only **reallocates between selection and transmission**. This is forced: both sides equal the total change in the firm-weighted mean trait, which telescopes trivially.

### Step 6 — when do the naive identities hold?

Each of (Q1), (Q2) holds **iff $C=0$**: iff second-period relative firm growth is uncorrelated (under 2020 firm shares) with first-period trait changes. Sufficient special cases: $\Delta z_1$ constant across parishes (a purely common first-year sales shock), or $w_2$ constant (uniform second-year firm growth). Generically $C\neq 0$ — the classical **time-non-additivity of the Price equation**: chained selection acts on each year's *updated* trait, the long window on the *initial* trait, and the wedge is "selection on the transmitted change" (a fitness–transmission covariance).

So, the identity does not really hold but there's a correction term that makes it exact. This term is the fitness transmission cross term, characterised by the covariance of the proportional relative fitness and the change of the trait through year $i \to j$ with $i \leq j$. 

The proof outline would go something like this, we have numerically shown that this identity does not hold. Then something must be missing and there should be a way for us to recover this. First, we have to re-weight: take the final shares, know that they are the initial shares titled by something called the relative fitness, so for any fixed vector we can express this "tiltness" using the $cov_{p(s)}(\frac{w}{\bar{W}},x)$. For selection this is a difference of means of the fixed initial trait under two different weightings: $S(s \to t) = \vec{Z_s}_{p(t)}-\vec{Z_s}_{p(s)}$.

Then, the chan can be decomposed exactly into two terms.

```python
# ==============================================================================
# Setup -- the yearly parish tables the numeric check below uses: firm counts,
# real sales, and the set of valid parishes in each year (same construction as
# the other notebooks, through ecuador_price/decomposition.py).
# ==============================================================================
import sys
from pathlib import Path
import numpy as np, pandas as pd

# The project root is the folder that holds ecuador_price/ (this notebook lives in notebooks/).
PROJECT_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents]
                    if (p / "ecuador_price").is_dir())
sys.path.insert(0, str(PROJECT_ROOT))
from ecuador_price import decomposition as dec

monthly = dec.monthly_table(dec.load_real_panel())
py = dec.parish_period_table(monthly, "yearly")            # one row per (parish, year)
valid_cells = dec.annual_valid_cells(monthly)              # firms>0 & real sales>$1,000
VALID = {y: dec.valid_parishes(py, str(y), valid_cells) for y in (2019, 2020, 2021)}
FIRMS = py.pivot(index="cod_parr", columns="year", values="firms")
SALES = py.pivot(index="cod_parr", columns="year", values="sales")
print({y: len(v) for y, v in VALID.items()}, "valid parishes per year")
```

```python
# ==============================================================================
# STEP K -- additivity of S and T: symbolic proof check + numeric verification
# ------------------------------------------------------------------------------
# S_(ns, nt, x) is the selection functional cov_{p(s)}(w/Wbar, x); with x = the
# first-period trait CHANGE it gives the cross term C (Lemma (b) applies to any
# fixed vector). T_(nt, zs, zt) is transmission = final-share mean of (zt - zs).
# (1) sympy, 3 fully general units: the naive sums are NOT identically the long
#     window; with -C (selection) / +C (transmission) they ARE; S+T is exactly
#     additive. (2) same on the common 2019/2020/2021 survivor set, to 1e-16.
# ==============================================================================
import sympy as sp

a = sp.symbols("a1 a2 a3", positive=True)   # firm counts 2019
b = sp.symbols("b1 b2 b3", positive=True)   # firm counts 2020
c = sp.symbols("c1 c2 c3", positive=True)   # firm counts 2021
x = sp.symbols("x1 x2 x3", real=True)       # ln sales 2019
y = sp.symbols("y1 y2 y3", real=True)       # ln sales 2020
u = sp.symbols("u1 u2 u3", real=True)       # ln sales 2021

def S_sym(ns, nt, xs):
    Ns, Nt = sum(ns), sum(nt)
    p = [n / Ns for n in ns]
    w = [t_ / s_ for s_, t_ in zip(ns, nt)]
    Wb = Nt / Ns
    xbar = sum(pi * xi for pi, xi in zip(p, xs))
    return sum(pi * (wi / Wb - 1) * (xi - xbar) for pi, wi, xi in zip(p, w, xs))

def T_sym(nt, zs, zt):
    Nt = sum(nt)
    return sum((ni / Nt) * (zti - zsi) for ni, zsi, zti in zip(nt, zs, zt))

S1_, S2_, S12_ = S_sym(a, b, x), S_sym(b, c, y), S_sym(a, c, x)
T1_, T2_, T12_ = T_sym(b, x, y), T_sym(c, y, u), T_sym(c, x, u)
C_ = S_sym(b, c, [yi - xi for xi, yi in zip(x, y)])       # cov_{p20}(w2/W2, dz1)
print("symbolic checks (3 fully general units):")
print("  S1+S2-S12 == 0 ?              ", sp.simplify(S1_ + S2_ - S12_) == 0)
print("  S1+S2-S12-C == 0 ?            ", sp.simplify(S1_ + S2_ - S12_ - C_) == 0)
print("  T1+T2-T12 == 0 ?              ", sp.simplify(T1_ + T2_ - T12_) == 0)
print("  T12-T1-T2-C == 0 ?            ", sp.simplify(T12_ - T1_ - T2_ - C_) == 0)
print("  (S12+T12)-(S1+T1)-(S2+T2)==0 ?", sp.simplify(S12_ + T12_ - S1_ - T1_ - S2_ - T2_) == 0)

# ---- numeric, on the common survivor set ---------------------------------------
common = sorted(VALID[2019] & VALID[2020] & VALID[2021])
Fj, Yj = FIRMS.loc[common], SALES.loc[common]

def S_num(ns, nt, xv):
    p = ns / ns.sum()
    w = nt / ns
    Wb = (p * w).sum()
    xb = (p * xv).sum()
    return float((p * (w / Wb - 1) * (xv - xb)).sum())

def T_num(nt, zs, zt):
    q = nt / nt.sum()
    return float((q * (zt - zs)).sum())

z19, z20, z21 = np.log(Yj[2019]), np.log(Yj[2020]), np.log(Yj[2021])
S1, S2  = S_num(Fj[2019], Fj[2020], z19), S_num(Fj[2020], Fj[2021], z20)
S12     = S_num(Fj[2019], Fj[2021], z19)
T1, T2  = T_num(Fj[2020], z19, z20), T_num(Fj[2021], z20, z21)
T12     = T_num(Fj[2021], z19, z21)
C       = S_num(Fj[2020], Fj[2021], z20 - z19)

print(f"\ncommon survivor set: {len(common)} parishes   (raw, unannualised)")
print(f"  cross term C = cov_p20(w2/W2, dz1)  = {C:+.6f}")
hdr = f"  {'':14s}{'window 1':>10s}{'window 2':>10s}{'sum':>10s}{'long':>10s}{'naive gap':>11s}{'residual':>11s}"
print("\n" + hdr)
print(f"  selection    {S1:+10.6f}{S2:+10.6f}{S1+S2:+10.6f}{S12:+10.6f}"
      f"{S1+S2-S12:+11.6f}{S1+S2-S12-C:+11.1e}")
print(f"  transmission {T1:+10.6f}{T2:+10.6f}{T1+T2:+10.6f}{T12:+10.6f}"
      f"{T1+T2-T12:+11.6f}{T1+T2-T12+C:+11.1e}")
print(f"  total S+T    {S1+T1:+10.6f}{S2+T2:+10.6f}{S1+T1+S2+T2:+10.6f}{S12+T12:+10.6f}"
      f"{(S1+T1+S2+T2)-(S12+T12):+11.1e}")
print("\n  selection residual = S1+S2-S12-C ;  transmission residual = T1+T2-T12+C")
print("  the naive gaps are equal and opposite (-C vs +C): the wedge only moves BETWEEN S and T.")
```

### What the check shows

- **The algebra is airtight.** sympy confirms all five statements for arbitrary firm counts and traits: neither $S$ nor $T$ is naively additive; each corrected identity is exact; the total $S+T$ is exactly additive. On the data every residual is $\sim 10^{-16}$.
- **The two naive gaps are equal and opposite** ($-C$ for selection, $+C$ for transmission, with $C=-0.00039$ here): the cross term is a pure *reallocation* between the two Price terms and never touches the total change $\langle z_{21}\rangle_{21}-\langle z_{19}\rangle_{19}$.
- **In these years the naive identities are decent numerical approximations** ($|C|\approx 4\times10^{-4}$, tiny next to $T_1=-0.155$ or $T_2=+0.212$) — but conceptually misleading for selection, where $|C|$ is ≈ 140% of the direct $S(19{\to}21)$. The reason is the striking empirical fact underneath: $S_1=+0.034$ and $S_2=-0.035$ **cancel almost perfectly** — over the two-year window net selection is $\approx 0$; the COVID concentration of firm formation toward the big parishes was fully undone by the rebound year. Transmission does the same in muted form: $-0.155$ then $+0.212$, netting $+0.057$.
- **The sign of $C$ is economically meaningful.** $C=\operatorname{cov}_{p(20)}(w_2/\bar W_2,\Delta z_1)<0$: parishes whose *sales fell* in the COVID year had *higher* relative firm growth in the rebound year — second-period fitness chased the first-period trait shock mean-revertingly. Correlation between one period's transmission and the next period's fitness is exactly what breaks time-additivity.

**Practical rule for the project:** the engine's per-window $S$ and $T$ can be *compared* across windows but not *summed* into a longer window — to aggregate over $k$ years, run the engine on the long window directly (as the headline windows do), or add the $k-1$ cross terms ($-C$'s to the selection chain, $+C$'s to the transmission chain). Summing the *totals* $S+T$ is always safe.
